## Reading from CSV files

```python
dataframes = {}

base_path = "/Volumes/workspace/bronze/source_system"

for folder in dbutils.fs.ls(base_path):
    if folder.isDir():

        folder_name = folder.name.rstrip("/")

        for file in dbutils.fs.ls(folder.path):
            if file.path.endswith(".csv"):

                file_name = file.name.replace(".csv", "")

                df = spark.read.csv(
                    file.path,
                    header=True,
                    inferSchema=True
                )

                dataframes[file_name] = df
```

## Write it to the bronze layer

```python
for table_name, df in dataframes.items():

    df.write \
        .format("delta") \
        .mode("overwrite") \
        .saveAsTable(f"bronze.{table_name}")
```
